```python
# ==========================================
# БИБЛИОТЕКА КЛАССОВ (Модель предметной области)
# ==========================================

class LegalService:
    """Базовый класс для любой юридической услуги."""
    def __init__(self, name, hourly_rate, hours_worked):
        self.name = name
        self.hourly_rate = hourly_rate      # Ставка в час
        self.hours_worked = hours_worked    # Отработанное время

    def __str__(self):
        return f"Услуга: {self.name} | Ставка: {self.hourly_rate} руб./ч | Часов: {self.hours_worked}"

    def calculate_final_fee(self, hours):
        """Базовый метод расчета стоимости (будет переопределен в дочерних классах)."""
        return hours * self.hourly_rate


class Litigation(LegalService):
    """Дочерний класс: Судебное представительство."""
    def __init__(self, name, hourly_rate, hours_worked, court_name):
        super().__init__(name, hourly_rate, hours_worked)
        self.court_name = court_name

    def __str__(self):
        # Используем строку родителя и добавляем свои данные
        return f"{super().__str__()} | Суд: {self.court_name}"

    def calculate_final_fee(self, hours):
        # Бизнес-логика: для суда минимальная оплата всегда за 10 часов
        actual_hours = max(10, hours)
        return actual_hours * self.hourly_rate


class Consulting(LegalService):
    """Дочерний класс: Юридическая консультация."""
    def __init__(self, name, hourly_rate, hours_worked, law_area):
        super().__init__(name, hourly_rate, hours_worked)
        self.law_area = law_area

    def __str__(self):
        return f"{super().__str__()} | Область права: {self.law_area}"

    def calculate_final_fee(self, hours):
        # Бизнес-логика: для консультации оплата строго по факту затраченных часов
        return hours * self.hourly_rate


class Client:
    """Базовый класс для клиента."""
    def __init__(self, client_name, case_number):
        self.client_name = client_name
        self.case_number = case_number

    def __str__(self):
        return f"Клиент: {self.client_name} | Дело №: {self.case_number}"


class RetainerClient(Client):
    """Дочерний класс: Клиент на абонентском обслуживании."""
    def __init__(self, client_name, case_number, retainer_fee):
        super().__init__(client_name, case_number)
        self.retainer_fee = retainer_fee

    def __str__(self):
        return f"{super().__str__()} | Абон. плата: {self.retainer_fee} руб./мес"


class CaseFile:
    """Класс-контейнер: Папка дела, объединяющая клиента и список услуг."""
    def __init__(self, client, services_list):
        self.client = client
        self.services_list = services_list

    def __str__(self):
        return f"📁 Досье: {self.client.client_name} (Дело № {self.client.case_number}) | Услуг в деле: {len(self.services_list)}"

    def calculate_total_cost(self):
        """Полиморфный расчет общей стоимости всех услуг по делу."""
        total_cost = 0
        for service in self.services_list:
            # Вызываем один и тот же метод, но каждый класс считает по-своему!
            total_cost += service.calculate_final_fee(service.hours_worked)
        return total_cost


# ==========================================
# main.py (СЦЕНАРИЙ ДЕМОНСТРАЦИИ РАБОТЫ)
# ==========================================

def main():
    print("=" * 50)
    print("1. Создание организационной структуры (Клиенты)")
    print("=" * 50)

    # Создаем клиента на абонентском обслуживании
    client1 = RetainerClient(client_name="ООО 'Ромашка'", case_number="А40-123/2024", retainer_fee=50000)
    print(client1)
    print()

    print("=" * 50)
    print("2. Формирование списка услуг по делу")
    print("=" * 50)

    # Услуга 1: Консультация (отработали 3 часа). Оплата будет за 3 часа.
    service1 = Consulting(
        name="Устная консультация",
        hourly_rate=5000,
        hours_worked=3,
        law_area="Корпоративное право"
    )

    # Услуга 2: Суд (отработали 5 часов). Но из-за правила "мин. 10 часов", оплата будет за 10 часов!
    service2 = Litigation(
        name="Представительство в суде",
        hourly_rate=8000,
        hours_worked=5,
        court_name="Арбитражный суд г. Москвы"
    )

    services_for_case = [service1, service2]

    for s in services_for_case:
        print(s)
    print()

    print("=" * 50)
    print("3. Создание досье дела и расчет стоимости")
    print("=" * 50)

    # Собираем всё в контейнер
    case_file = CaseFile(client=client1, services_list=services_for_case)
    print(case_file)

    # Запускаем полиморфный расчет
    final_cost = case_file.calculate_total_cost()

    print("-" * 50)
    print(f"💰 Итоговая стоимость услуг по делу: {final_cost} руб.")
    print("(Проверка: Консультация 3ч * 5000 = 15 000. Суд 10ч(мин) * 8000 = 80 000. Итого: 95 000)")


# Запуск главного сценария
if __name__ == "__main__":
    main()
```
