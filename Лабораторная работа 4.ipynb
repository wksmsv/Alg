```python
# Задание №2. Блок ОСНОВЫ HASHLIB
#
# Выполнила: Самусева В.А.
# Группа: циб-251

import hashlib

def get_sha256_hash(s: str) -> str:
    """
    Возвращает SHA-256 хеш переданной строки в hex-формате.
    """
    # Кодируем строку в байты (UTF-8) и вычисляем хеш
    return hashlib.sha256(s.encode('utf-8')).hexdigest()

# --- Проверка на тестовых данных ---
if __name__ == "__main__":
    test_input = "Python"
    result = get_sha256_hash(test_input)
    expected = "18885f27b5af9012df19e496460f9294d5ab76128824c6fdc565542143726f3e"

    print(f"Input: {test_input}")
    print(f"Output: {result}")
    print(f"Тест пройден: {result == expected}")
```
